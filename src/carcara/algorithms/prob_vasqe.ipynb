```python
import numpy as np

def modified_softmax(A, tau):
    # Garante que a entrada seja um array do NumPy
    A = np.array(A, dtype=float)
    
    # 1. Aplica o sinal trocado e a escala da temperatura
    exponent = -A / tau
    
    # Truque de Estabilidade Numérica:
    # Subtraímos o valor máximo dos expoentes antes de calcular a exponencial.
    # Isso evita que e^x resulte em 'inf' (overflow) se os valores forem muito pequenos/negativos.
    stable_exponent = exponent - np.max(exponent)
    
    # 2. Calcula a exponencial de cada elemento
    exps = np.exp(stable_exponent)
    
    # 3. Normaliza para que a soma seja igual a 1
    probabilities = exps / np.sum(exps)
    
    return probabilities

# Exemplo com 5 valores não-negativos
A = [1.5, 2.0, 0.5, 4.0, 1.0]
tau = 2.0  # Temperatura

probabilidades = modified_softmax(A, tau)

print(f"Vetor original A: {A}")
print(f"Probabilidades:   {probabilidades}")
print(f"Soma das probabs: {np.sum(probabilidades):.1f}")

escolha = np.random.choice(A, p=probabilidades)
print(f"Valor selecionado diretamente: {escolha}")
```
